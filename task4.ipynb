```python
# Install library
!pip install mlxtend

# Import libraries
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

# -----------------------------------
# Retail Transaction Dataset
# -----------------------------------
transactions = [
['milk','bread','butter'],
['beer','bread'],
['milk','bread','beer','butter'],
['bread','butter'],
['milk','bread','butter','beer'],
['milk','bread'],
['bread','beer'],
['milk','butter'],
['bread','butter','beer'],
['milk','bread','butter'],
['milk','bread','eggs'],
['bread','eggs'],
['milk','bread','butter','eggs'],
['milk','butter','cheese'],
['bread','butter','cheese'],
['milk','bread','cheese'],
['milk','eggs','cheese'],
['bread','butter','jam'],
['milk','bread','jam'],
['bread','jam'],
['milk','butter','jam'],
['bread','butter','eggs'],
['milk','bread','butter','jam'],
['milk','bread','cheese','eggs'],
['bread','cheese'],
['milk','bread','butter','cheese'],
['bread','eggs','jam'],
['milk','bread','jam','eggs'],
['milk','bread','butter','eggs'],
['bread','butter','cheese','jam'],
['milk','cheese'],
['bread','butter'],
['milk','bread','eggs','jam'],
['milk','butter','cheese','jam'],
['bread','butter','eggs','cheese'],
['milk','bread','butter','cheese','eggs'],
['milk','bread','butter'],
['bread','eggs','cheese'],
['milk','bread','cheese'],
['milk','butter','eggs']
]

print("Sample Transactions:")
print(transactions[:10])

# -----------------------------------
# One Hot Encoding
# -----------------------------------
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_array, columns=te.columns_)

print("\nEncoded Dataset:")
print(df.head())

# -----------------------------------
# APRIORI
# -----------------------------------
print("\n===== Apriori Frequent Itemsets =====")

apriori_items = apriori(df, min_support=0.2, use_colnames=True)

print(apriori_items.sort_values(by="support", ascending=False))

# -----------------------------------
# Apriori Association Rules
# -----------------------------------
rules_apriori = association_rules(apriori_items, metric="confidence", min_threshold=0.6)

print("\n===== Apriori Association Rules =====")

print(rules_apriori[['antecedents','consequents','support','confidence','lift']])

# -----------------------------------
# FP Growth
# -----------------------------------
print("\n===== FP Growth Frequent Itemsets =====")

fp_items = fpgrowth(df, min_support=0.2, use_colnames=True)

print(fp_items.sort_values(by="support", ascending=False))

# -----------------------------------
# FP Growth Association Rules
# -----------------------------------
rules_fp = association_rules(fp_items, metric="confidence", min_threshold=0.6)

print("\n===== FP Growth Association Rules =====")

print(rules_fp[['antecedents','consequents','support','confidence','lift']])

# -----------------------------------
# Comparison
# -----------------------------------
print("\n===== Algorithm Comparison =====")

print("Apriori Itemsets:", len(apriori_items))
print("FP Growth Itemsets:", len(fp_items))

print("Apriori Rules:", len(rules_apriori))
print("FP Growth Rules:", len(rules_fp))
```
