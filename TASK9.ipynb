```python
import numpy as np

class Perceptron:
    def __init__(self, input_dim, learning_rate=0.1):
        self.weights = np.zeros(input_dim + 1)  # Including bias
        self.lr = learning_rate

    def predict(self, X):
        linear_output = np.dot(X, self.weights[1:]) + self.weights[0]
        return np.where(linear_output >= 0, 1, 0)

    def fit(self, X, y, epochs=10):
        for _ in range(epochs):
            for xi, yi in zip(X, y):
                y_pred = self.predict(xi)
                error = yi - y_pred
                self.weights[1:] += self.lr * error * xi
                self.weights[0] += self.lr * error  # bias update
        print("Training complete. Weights:", self.weights)

# Example usage
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1])  # AND gate

perceptron = Perceptron(input_dim=2)
perceptron.fit(X, y, epochs=10)
print("Predictions for input [1, 1]:", perceptron.predict(np.array([1, 1])))
```

```python
# Logical gate datasets
X_and = np.array([[0,0], [0,1], [1,0], [1,1]])
y_and = np.array([0, 0, 0, 1])

X_or = np.array([[0,0], [0,1], [1,0], [1,1]])
y_or = np.array([0, 1, 1, 1])

X_not = np.array([[0], [1]])
y_not = np.array([1, 0])

def train_gate(X, y):
    p = Perceptron(input_dim=X.shape[1])
    p.fit(X, y, epochs=10)
    return p

# Train AND gate
and_gate = train_gate(X_and, y_and)
print("AND Gate Predictions:", and_gate.predict(X_and))

# Train OR gate
or_gate = train_gate(X_or, y_or)
print("OR Gate Predictions:", or_gate.predict(X_or))

# Train NOT gate
not_gate = train_gate(X_not, y_not)
print("NOT Gate Predictions:", not_gate.predict(X_not))
```

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

class MLP:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.1):
        self.lr = learning_rate
        self.W1 = np.random.randn(input_size, hidden_size) * 0.1
        self.b1 = np.zeros((1, hidden_size))
        self.W2 = np.random.randn(hidden_size, output_size) * 0.1
        self.b2 = np.zeros((1, output_size))

    def forward(self, X):
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = sigmoid(self.z1)
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = sigmoid(self.z2)
        return self.a2

    def backward(self, X, y, output):
        error = y - output
        delta2 = error * sigmoid_derivative(output)
        delta1 = delta2.dot(self.W2.T) * sigmoid_derivative(self.a1)
        # Update weights
        self.W2 += self.a1.T.dot(delta2) * self.lr
        self.b2 += np.sum(delta2, axis=0, keepdims=True) * self.lr
        self.W1 += X.T.dot(delta1) * self.lr
        self.b1 += np.sum(delta1, axis=0, keepdims=True) * self.lr

    def train(self, X, y, epochs=10000):
        for epoch in range(epochs):
            output = self.forward(X)
            self.backward(X, y, output)
            if epoch % 2000 == 0:
                loss = np.mean((y - output) ** 2)
                print(f"Epoch {epoch}, Loss: {loss:.4f}")

    def predict(self, X):
        output = self.forward(X)
        return np.round(output)

# XOR data
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

# Train the network
np.random.seed(42)
mlp = MLP(input_size=2, hidden_size=2, output_size=1, learning_rate=0.5)
mlp.train(X, y, epochs=10000)

# Predictions
print("Predictions for XOR input:")
print(mlp.predict(X))
```
